# Cross-Platform Algorithmic Throughput Analysis

This notebook analyzes cross-platform algorithmic throughput for various processing paradigms, computes cross-platform averages, and normalizes performance relative to **Legacy Paradigm A**.

## Table 1: Cross-Platform Throughput (FPS) for Various Processing Paradigms

```python
import numpy as np
import pandas as pd

# Table 1: Cross-Platform Algorithmic Throughput (k/s)
# Figures taken strictly from the markdown table
platforms = ["VS Code Preview", "Chrome", "Firefox", "Android Chrome", "Android Firefox", "Android Comet"]

data = {
    "Algorithm": [
        "Legacy Paradigm A",
        "Legacy Paradigm B",
        "Stateless Perlin",
        "Stateless Worley",
        "FPS-R SM",
        "FPS-R TM",
        "FPS-R QS",
        "FPS-R BD (default 3 streams, blocksize 64)",
    ],
    "VS Code Preview": [2483, 2298, 870, 600, 1385, 1834, 872, 350],
    "Chrome": [2953, 2587, 1016, 691, 1747, 2135, 956, 390],
    "Firefox": [7710, 7300, 405, 226, 1092, 1838, 678, 328],
    "Android Chrome": [3010, 2724, 1076, 725, 1873, 2296, 1125, 423],
    "Android Firefox": [3550, 3360, 320.3, 183.8, 855.3, 1213, 435.5, 245.6],
    "Android Comet": [3188, 2583, 871.8, 590.1, 1537.1, 1923.4, 930.2, 308.3],
}

df_throughput = pd.DataFrame(data).set_index("Algorithm")

# Display Table 1 formatted with tildes and commas
table_1 = df_throughput.map(lambda x: f"~{x:,}")
table_1
```

---
## Table 2: Average Throughput Across Platforms

```python
# Table 2: Averaged Algorithmic Throughput (k/s)
# Calculated as the mean across the 4 platform columns
df_averaged = pd.DataFrame(
    {"Average Performance": df_throughput[platforms].mean(axis=1)}
)

# Standardize FPS-R BD label for summary tables
df_averaged.index = df_averaged.index.map(
    lambda name: "FPS-R BD" if "FPS-R BD" in name else name
)

# Format to 2 decimal places with tildes matching markdown
table_2 = df_averaged.map(lambda x: f"~{x:,.2f}")
table_2
```

---
## Table 3: Normalized Performance Relative to Legacy Paradigm A

```python
# Table 3: Performance Relative to Legacy Paradigm A Across Platforms
# R_{i, j} = performance(i, j) / Legacy_A(j)
legacy_a_baseline = df_throughput.loc["Legacy Paradigm A", platforms]
df_relative = df_throughput[platforms].div(legacy_a_baseline, axis=1)

# Compute the median ratio across the 4 platforms
df_relative["Median ratio"] = df_relative[platforms].median(axis=1)

# Simplify index label for FPS-R BD
df_relative.index = df_relative.index.map(
    lambda name: "FPS-R BD" if "FPS-R BD" in name else name
)


# Format each cell as "ratio / percentage%"
def format_ratio_pct(val):
    percent = val * 100
    percent_flag = False  # This variable is defined but not used; consider removing it if unnecessary
    return_str = f"{val:.3f}"
    if percent_flag:
        return_str += f" / {percent:.1f}%"
    return return_str


table_3 = df_relative.map(format_ratio_pct)
table_3
```

```python
# Display ranked format_ratio_pct for table 3
ranked_median_ratios = df_relative["Median ratio"].sort_values(ascending=False)
print(ranked_median_ratios)

# print ranked_median_ratios in a percentage format
print(ranked_median_ratios.apply(lambda x: f"{x:.1%}"))
```

---
## Generate Markdown Table for Cross-Platform Throughput

```python
# generate a markdown table for cross-platform throughput
# starting with headers:
#     | Algorithm | VS Code Preview | Chrome | Firefox | Android Chrome |
#     | :--- | ---: | ---: | ---: | ---: |

# generate the table with the formatted values
def df_to_markdown(df):
    headers = ["Algorithm"] + list(df.columns)
    alignments = [":---"] + ["---:"] * len(df.columns)
    header_line = "| " + " | ".join(headers) + " |"
    separator_line = "| " + " | ".join(alignments) + " |"
    rows = [
        "| " + " | ".join([str(idx)] + [str(val) for val in row]) + " |"
        for idx, row in df.iterrows()
    ]
    return "\n".join([header_line, separator_line] + rows)

print("### Table 1: Cross-Platform Throughput (FPS) for Various Processing Paradigms")
print(df_to_markdown(table_1))
print()

print("### Table 2: Average Throughput Across Platforms")
print(df_to_markdown(table_2))
print()

print("### Table 3: Normalized Performance Relative to Legacy Paradigm A")
print(df_to_markdown(table_3))
```
